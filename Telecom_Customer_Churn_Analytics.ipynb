<h3>Load Dataset</h3>

<p style="margin-bottom: 10px; color: #431407 !important; line-height: 1.6;">
    Working directly with the massive 1 million row telecom churn dataset introduces memory constraints and slow rendering times during rapid prototyping. By applying Stratified Sampling, we downsample the dataset to 3,000 records without distorting the underlying target churn distribution. This reduction streamlines EDA operations and speeds up iterative visual analysis.
  </p>


</div>

```python
# Load dataset
df = pd.read_csv("customer_churn.csv")

# Reduce dataset to 3,000 rows while preserving churn proportions
df = (
    df.groupby("churn", group_keys=False)
    .apply(
        lambda x: x.sample(n=int(3001 * len(x) / len(df)), random_state=42)
    )
    .reset_index(drop=True)
)
print(f"Downsampled Dataset Shape: {df.shape}")

# Display initial rows
df.head(3)
```

<div style="background-color: #eef2ff !important; padding: 16px 20px; border-radius: 8px; border: 1px solid #c7d2fe !important; border-left: 6px solid #4f46e5 !important; margin: 20px 0 10px 0;">

 <h3 style="color: #1d4ed8 !important; margin-top: 0; font-size: 28px; margin-bottom: 12px; font-weight: 600; text-align: center;">🧹<strong>Data Cleaning & Preprocessing</strong></h3>
 <p style="margin-bottom: 10px; color: #431407 !important; line-height: 1.6;">
    This step executes targeted data transformations to achieve a clean, model-ready dataset:
  </p>

  <ul style="line-height: 1.7; color: #431407 !important; margin-bottom: 0; padding-left: 20px;">
    <li><strong>Type Casting:</strong> Converts <code>signup_date</code> from text string (<code>object</code>) to proper <code>datetime</code> format.</li>
    <li><strong>Zero Imputation:</strong> Replaces missing values in <code>num_complaints</code> with <code>0</code> (assuming no complaints were logged).</li>
    <li><strong>Median Imputation:</strong> Fills missing values in numerical columns (<code>annual_income</code>, <code>credit_score</code>, <code>avg_monthly_gb</code>, and <code>customer_satisfaction</code>) using medians.</li>
    <li><strong>Standardization:</strong> This step automatically cleans up text entries in all text (<code>object</code>) columns to prevent duplicate categories caused by inconsistent typing:.</li>
      <li><strong>Duplicate Removal:</strong> This step removes all the duplicate rows.</li>
        <li><strong>Droping Non-Predictive columns:</strong> This step delete those columns which are not contributing to the analysis.</li>
  </ul>
</div>

<div style="background-color: #eef2ff !important; padding: 16px 20px; border-radius: 8px; border: 1px solid #c7d2fe !important; border-left: 6px solid #4f46e5 !important; margin: 20px 0 10px 0;">
    <h4 style="color: #1d4ed8 !important; margin-top: 0; font-size: 28px; margin-bottom: 12px; font-weight: 600;">Type Casting</h4>
     <p style="margin-bottom: 10px; color: #431407 !important; line-height: 1.6;">
    Here we create a copy of dataset (df_clean) which provide separate workspace where we can edit, clean, and modify data freely without any warnings or risk of making mistakes. So that our original dataset (df) stays safe and untouched in memory.
  </p>
</div>

```python
# Create a clean copy to avoid SettingWithCopy warning
df_clean = df.copy()

# Convert signup_date to proper datetime format

df_clean["signup_date"] = pd.to_datetime(df_clean["signup_date"],format="mixed" ,errors="coerce")

print(f"The changed data type of signup_date is :{df_clean["signup_date"].dtype}")
```

<div style="background-color: #fff5f0 !important; color: #4a1d17 !important; padding: 5px; border-radius: 8px; border: 1px solid #e2e8f0 !important; border-left: 5px solid #3b82f6 !important; font-family: sans-serif; margin-top: 15px;">
     <h3 style="margin-top: 0; margin-bottom: 0; line-height: 1.6; color: #166534 !important; font-weight: 600;">
    ✅ Result</h3>
      <h4 style="color: #1d4ed8 !important; margin-top: 0; font-size: 18px; margin-bottom: 12px; font-weight: 600;">
    Data Type Update:
  </h4>
   <p> The signup_date column was successfully converted from string (object) into datetime64[ns] timestamp</p>
</div>

<div style="background-color: #eef2ff !important; padding: 16px 20px; border-radius: 8px; border: 1px solid #c7d2fe !important; border-left: 6px solid #4f46e5 !important; margin: 20px 0 10px 0;">
    <h4 style="color: #1d4ed8 !important; margin-top: 0; font-size: 28px; margin-bottom: 12px; font-weight: 600;">Handling Missing Values</h4>
     <p style="margin-bottom: 10px; color: #431407 !important; line-height: 1.6;"><strong>Zero Imputation (num_complaints):</strong> Replacing missing values with 0 accurately reflects zero logged customer complaints.</p>
    <p style="margin-bottom: 10px; color: #431407 !important; line-height: 1.6;"><strong>Median Imputation (Numerical Features):</strong> Continuous variables (annual_income, credit_score, avg_monthly_gb, customer_satisfaction) often contain skewed distributions or extreme values. Median provides a robust central estimate that minimizes bias from outliers compared to the mean.</p>
</div>

```python
# Behavioral count imputation (Missing complaints = 0 complaints logged)
df_clean["num_complaints"] = df_clean["num_complaints"].fillna(0)

# Median imputation for numerical columns
num_cols = [
    "annual_income",
    "credit_score",
    "avg_monthly_gb",
    "customer_satisfaction",
]
for col in num_cols:
    median_value = df_clean[col].median()
    df_clean[col] = df_clean[col].fillna(median_value)

# Verification & Post-Cleaning Audit Table

remaining_nulls = df_clean.isnull().sum().sum()

print("Data Cleaning Completed Successfully!")
print(f"Total Remaining Nulls Across Dataset: {remaining_nulls}\n")
```

<div style="background-color: #fff5f0!important; color: #4a1d17 !important; padding: 20px; border-radius: 8px; border: 1px solid #e2e8f0 !important; border-left: 5px solid #3b82f6 !important; font-family: sans-serif; margin-top: 15px;">

 <h3 style="margin-top: 0; margin-bottom: 0; line-height: 1.6; color: #166534 !important; font-weight: 600;">
    ✅ Result
  <h/3>
<h4 style="color: #1d4ed8 !important; margin-top: 0; font-size: 18px; margin-bottom: 12px; font-weight: 600;">
     Missing Value Imputation & Quality Audit
  </h4>
  
  <p style="margin-bottom: 12px; line-height: 1.6;">
    This step removes all missing values (<code>NaN</code>) so machine learning models can run without errors:
  </p>

  <ul style="line-height: 1.7; color: #334155 !important; margin-bottom: 12px; padding-left: 20px;">
    <li style="margin-bottom: 8px;">
      <strong>Complaints (<code>num_complaints</code>):</strong> Blank values are filled with <code>0</code>, assuming no complaints were logged by the customer.
    </li>
    <li style="margin-bottom: 8px;">
      <strong>Numerical Columns (Income, Credit Score, GB Usage, Satisfaction):</strong> Missing values are filled with the <strong>median</strong> (middle value) so extreme outliers do not distort the dataset.
    </li>
    <li>
      <strong>Verification Audit:</strong> Sums all missing entries across the dataset to confirm a <strong>0 remaining null</strong> count.
    </li>
  </ul>


  </div>

<div style="background-color: #eef2ff !important; padding: 16px 20px; border-radius: 8px; border: 1px solid #c7d2fe !important; border-left: 6px solid #4f46e5 !important; margin: 20px 0 10px 0;">
    <h4 style="color: #1d4ed8 !important; margin-top: 0; font-size: 28px; margin-bottom: 12px; font-weight: 600;">Standardization</h4>
     <p style="margin-bottom: 10px; color: #431407 !important; line-height: 1.6;"><strong>Find Text Columns:</strong> Select_dtypes(include="object") automatically scans the dataset and collects every column containing text.</p>
    <p style="margin-bottom: 10px; color: #431407 !important; line-height: 1.6;"><strong>Trim Spaces(.str.strip()):</strong> Removes accidental spaces at the start or end of text.</p>
    <p style="margin-bottom: 10px; color: #431407 !important; line-height: 1.6;"><strong>Title(.str.title()):</strong> Capitalizes the first letter of each word.</p>
    <p style="margin-bottom: 10px; color: #431407 !important; line-height: 1.6;"><strong>Format Spaces(.str.replace()):</strong> Replaces inner spaces with underscores.</p>
</div>

```python
# Identify all categorical columns
string_cols = df_clean.select_dtypes(include=["object"]).columns

# Apply text standardization across all string columns
for col in string_cols:
    df_clean[col] = (
        df_clean[col]
        .astype(str)  # keep it as string
        .str.strip()  # Remove leading/trailing whitespaces
        .str.title()  # Convert first letter to caps
        .str.replace(" ", "_")  # Replace spaces with underscores
    )

print(f"Standardized {len(string_cols)} categorical columns!")
```

 <h3 style="margin-top: 0; margin-bottom: 0; line-height: 1.6; color: #166534 !important; font-weight: 600;">
    ✅ Result
  <h/3>
 <h4 style="color: #1d4ed8 !important; margin-top: 0; font-size: 18px; margin-bottom: 12px; font-weight: 600;">
    Text Standardization
  </h4>
  
  <p style="margin-bottom: 12px; line-height: 1.6; color: #334155 !important;">
    This step automatically cleans up text entries in all text (<code>object</code>) columns to prevent duplicate categories caused by inconsistent typing:
  </p>
   <p style="margin-bottom: 12px; line-height: 1.6; color: #334155 !important;">
   All 7 categorical columns are standardized
  </p>

</div>

    <h4 style="color: #1d4ed8 !important; margin-top: 0; font-size: 28px; margin-bottom: 12px; font-weight: 600;">Outlier Analysis & Handling Strategy</h4>
     <p style="margin-bottom: 10px; color: #431407 !important; line-height: 1.6;">Outliers are extreme or unusual data points that fall far outside the expected range of the dataset. Identifying these values is crucial during EDA because unusually high or low entries. Using box plots and Interquartile Range (IQR) filtering helps us distinguish whether these extreme records reflect legitimate high-value customers or data collection errors.</p>
     <p style="margin-bottom: 10px; color: #431407 !important; line-height: 1.6;">In this step we are going to identify and inspect outliers in continuous numerical columns (`annual_income`, `monthlycharges`, `totalcharges`, `avg_monthly_gb`) using the<strong> IQR</strong> method ($Q_1 - 1.5 \times \text{IQR}$ to $Q_3 + 1.5 \times \text{IQR}$).</p>
</div>

```python
# Identify and inspect outliers in continuous numerical columns using the IQR method
continuous_cols = ["annual_income", "monthlycharges", "totalcharges", "avg_monthly_gb"]

outlier_summary = []

for col in continuous_cols:
    if col in df_clean.columns:
        # Drop NaN values for accurate IQR calculation
        col_data = df_clean[col].dropna()
        
        Q1 = col_data.quantile(0.25)
        Q3 = col_data.quantile(0.75)
        IQR = Q3 - Q1
        
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        # Count values outside the IQR bounds
        outlier_count = ((col_data < lower_bound) | (col_data > upper_bound)).sum()
        outlier_pct = (outlier_count / len(df_clean)) * 100
        
        outlier_summary.append({
            "Feature": col,
            "Outlier Count": outlier_count,
            "Percentage (%)": round(outlier_pct, 2)
        })

# Display as DataFrame
outlier_df = pd.DataFrame(outlier_summary)
outlier_df
```

 <p style="margin-bottom: 12px; line-height: 1.6; color: #1e293b !important;">
    Feature engineering creates new columns from raw data to help the analysis of factors affecting churn more accurately:
  </p>
  <p style="margin-bottom: 12px; line-height: 1.6; color: #1e293b !important;">
    In this Telecom Customer Churn project following columns are going to create for better analysis.

  <ul style="line-height: 1.7; color: #334155 !important; margin-bottom: 12px; padding-left: 20px;">
    <li style="margin-bottom: 8px;">
      <strong>Tenure in Years (tenure_years):</strong> Converts tenure from months into years, making it easy to group short-term vs. long-term customers.
    </li>
    <li style="margin-bottom: 8px;">
      <strong>Monthly Complaint Rate (complaint_rate_per_month):</strong> Divides complaints by tenure. This separates new unhappy users from loyal customers who logged a single issue years ago.
    </li>
    <li>
      <strong>Low Satisfaction Flag (is_low_satisfaction):</strong> Creates a simple 1 (Yes) or 0 (No) flag for users rating satisfaction 2 or lower to instantly flag high-risk accounts.

<div style="background-color: #eef2ff !important; padding: 16px 20px; border-radius: 8px; border: 1px solid #c7d2fe !important; border-left: 6px solid #4f46e5 !important; margin: 20px 0 10px 0;">
    <h4 style="color: #1d4ed8 !important; margin-top: 0; font-size: 28px; margin-bottom: 12px; font-weight: 600;">Calculate Tenure in Years</h4>
    <p style="margin-bottom: 10px; color: #431407 !important; line-height: 1.6;">Converting customer tenure from raw months into years directly aligns data analysis with standard industry contract lifecycles and retention milestones. Raw monthly figures (such as 23 versus 24 months) introduce unnecessary granularity and noise, whereas measuring in years smooths minor month-to-month variance while preserving the core long-term retention trend. This annual framing not only makes visual charts and cohort heatmaps significantly cleaner for executive review, but it also creates intuitive features that can be easily combined with financial metrics (like total charges or annual income).</p>
     <p style="margin-bottom: 10px; color: #431407 !important; line-height: 1.6;">Converting customer tenure from raw months into years (tenure / 12).</p>    
</div>

```python
# Feature 1: Convert tenure from raw months into years
df_clean["tenure_years"] = (df_clean["tenure"] / 12).round(1)

# Verification
print("Feature 1 Created: 'tenure_years'")
df_clean[["tenure", "tenure_years"]].head()
```

<div style="background-color: #eef2ff !important; padding: 16px 20px; border-radius: 8px; border: 1px solid #c7d2fe !important; border-left: 6px solid #4f46e5 !important; margin: 20px 0 10px 0;">
    <h4 style="color: #1d4ed8 !important; margin-top: 0; font-size: 28px; margin-bottom: 12px; font-weight: 600;">Monthly Complaint Rate</h4>
     <p style="margin-bottom: 10px; color: #431407 !important; line-height: 1.6;">Dividing total complaints by customer tenure ($\frac{\text{num\_complaints}}{\text{tenure} + 1}$).</p>
    <p style="margin-bottom: 10px; color: #431407 !important; line-height: 1.6;">This normalizes complaints so we can isolate recent, highly dissatisfied signups.</p>
</div>

```python
# Feature 2: complaint frequency per month
# (+1 handles 0-month tenure to avoid division by zero errors)
df_clean["complaint_rate_per_month"] = (df_clean["num_complaints"] / (df_clean["tenure"] + 1)).round(3) # (+1 handles 0-month tenure to avoid division by zero errors)

# Verification
print(" Feature 2 Created: 'complaint_rate_per_month'")
df_clean[["tenure", "num_complaints", "complaint_rate_per_month"]].head()
```

<div style="background-color: #eef2ff !important; padding: 16px 20px; border-radius: 8px; border: 1px solid #c7d2fe !important; border-left: 6px solid #4f46e5 !important; margin: 20px 0 10px 0;">
    <h4 style="color: #1d4ed8 !important; margin-top: 0; font-size: 28px; margin-bottom: 12px; font-weight: 600;">Low Satisfaction Risk Flag</h4>
    <p style="margin-bottom: 10px; color: #431407 !important; line-height: 1.6;">Creating a Low Satisfaction Risk Flag transforms continuous or multi-grade customer feedback ratings into an explicit, immediate indicator for churn prediction. When an algorithm evaluates raw feedback numbers, it treats every score change incrementally; however, customer churn behavior typically triggers sharply below specific threshold scores. By establishing a clear binary threshold (assigning a 1 for high-risk, low-satisfaction scores and a 0 otherwise), we create an explicit "alarm bell" feature that highlights critical customer dissatisfaction. This simplified signal eliminates ambiguity, allows predictive models to capture non-linear churn risks faster during training, and enables immediate automated triggers for customer success teams to deploy targeted retention interventions.</p>
     <p style="margin-bottom: 10px; color: #431407 !important; line-height: 1.6;">Creating a simple binary flag ($1$ if satisfaction $\le 2$, else $0$).</p>
    
</div>

```python
# Feature 3: Create binary flag for high-risk low satisfaction
df_clean["is_low_satisfaction"] = (df_clean["customer_satisfaction"] <= 2).astype(int)

# Verification
print("Feature 3 Created: 'is_low_satisfaction'")
df_clean[["customer_satisfaction", "is_low_satisfaction"]].head()
```
